```python
import re
import pandas as pd

def extract_amount(text):
    if not isinstance(text, str) or len(text.strip()) == 0:
        return None
    
    keyword_match = re.findall(
        r'(?:total|amount due|amt due|grand total|sum|tota)[^\d]{0,30}(\d{1,6}[.,]\d{2})',
        text, re.IGNORECASE
    )
    if keyword_match:
        for m in keyword_match:
            try:
                val = float(m.replace(',', '.'))
                if 1.0 <= val <= 10000.0:
                    return round(val, 2)
            except:
                continue
    
    blocklist = ['vat', 'vatable', 'subtotal', 'sub total', 'cash', 'change', 
                 'discount', 'zero rated', 'exempt', 'zero-rated']
    
    all_numbers = list(re.finditer(r'\d{1,6}[.,]\d{2}', text))
    candidates = []
    
    for match in all_numbers:
        num_str = match.group(0)
        start_pos = match.start()
        preceding = text[max(0, start_pos - 40):start_pos].lower()
        
        if any(word in preceding for word in blocklist):
            continue
        
        try:
            val = float(num_str.replace(',', '.'))
            if 1.0 <= val <= 10000.0:
                candidates.append(val)
        except:
            continue
    
    return round(max(candidates), 2) if candidates else None


def extract_date(text):
    if not isinstance(text, str):
        return None
    
    patterns = [r'(\d{2}/\d{2}/\d{4})', r'(\d{4}-\d{2}-\d{2})']
    
    for pat in patterns:
        match = re.search(pat, text)
        if match:
            date_str = match.group(1)
            try:
                if '/' in date_str:
                    m, d, y = map(int, date_str.split('/'))
                else:
                    y, m, d = map(int, date_str.split('-'))
                
                if 1 <= m <= 12 and 1 <= d <= 31 and 2020 <= y <= 2030:
                    return f"{y}-{m:02d}-{d:02d}"
            except:
                continue
    return None


df = pd.read_csv("../outputs/metrics/easyocr_results.csv")
ground_truth = pd.read_csv("../data/annotated/ground_truth.csv")

results = []

for _, row in df.iterrows():
    extracted_amount = extract_amount(row.get('extracted_text', ''))
    extracted_date = extract_date(row.get('extracted_text', ''))
    
    results.append({
        "filename": row["filename"],
        "condition": row["condition"],
        "extracted_amount": extracted_amount,
        "extracted_date": extracted_date,
        "extracted_text": row.get("extracted_text", "")
    })

final_df = pd.DataFrame(results)
final_df = pd.merge(ground_truth, final_df, on=['filename', 'condition'])

final_df['amount_correct'] = abs(final_df['amount'] - final_df['extracted_amount'].fillna(0)) < 2.0

print("="*80)
print("               IMPROVED TEXT-BASED POSTPROCESSING")
print("="*80)
print(final_df[['filename', 'amount', 'extracted_amount', 'amount_correct', 'extracted_date']])

print(f"\nAmount Accuracy: {final_df['amount_correct'].mean()*100:.1f}%")
print(f"Successfully Extracted: {final_df['extracted_amount'].notna().sum()}/{len(final_df)}")

final_df.to_csv("../outputs/metrics/final_extraction.csv", index=False)
print("\n✅ Results saved to outputs/metrics/final_extraction.csv")
```
