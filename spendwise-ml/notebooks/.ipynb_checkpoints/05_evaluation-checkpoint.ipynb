```python
import pandas as pd

df = pd.read_csv("../outputs/metrics/final_extraction.csv")
ground_truth = pd.read_csv("../data/annotated/ground_truth.csv")

results = pd.merge(ground_truth, df[['filename', 'extracted_text', 'extracted_amount']], on='filename')

results['amount_correct'] = abs(results['amount'] - results['extracted_amount']) < 1.0
results['amount_extracted'] = results['extracted_amount'].notna()

total = len(results)
amount_correct_count = results['amount_correct'].sum()
amount_extracted_count = results['amount_extracted'].sum()

print("="*60)
print("          SPENDWISE OCR EVALUATION SUMMARY")
print("="*60)
print(f"Total Receipts          : {total}")
print(f"Amount Successfully Extracted : {amount_extracted_count}/{total} ({amount_extracted_count/total*100:.1f}%)")
print(f"Amount Correct          : {amount_correct_count}/{total} ({amount_correct_count/total*100:.1f}%)")
print(f"Needs User Correction   : {total - amount_correct_count}/{total} ({(total - amount_correct_count)/total*100:.1f}%)")
print("="*60)

print("\nPerformance by Condition:")
print(results.groupby('condition').agg({
    'amount_correct': ['mean', 'count'],
    'amount_extracted': 'mean'
}).round(3))

results.to_csv("../outputs/metrics/evaluation_results.csv", index=False)
print("\n✅ Detailed results saved to: outputs/metrics/evaluation_results.csv")
```
